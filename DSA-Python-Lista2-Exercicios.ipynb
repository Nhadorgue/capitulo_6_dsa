#### Exercício 1

Escreva uma função que receba os valores de três lados de um triângulo e o classifique como "Equilátero" (todos os lados iguais), "Isósceles" (dois lados iguais) ou "Escaleno" (todos os lados diferentes).

```python
# Solução
def teste(ladoA: int, ladoB: int, ladoC: int):
    if ladoA != ladoB & ladoA != ladoC & ladoB != ladoC:
        print("Triângulo Escaleno")

    elif (ladoA == ladoB & ladoA != ladoC) or \
         (ladoC == ladoB & ladoA != ladoC):
        print("Triângulo Isóceles")

    elif (ladoA == ladoB & ladoA == ladoC):
        print("Triângulo Equilátero")

teste(int(input("lado 1:")), int(input("lado 2:")), int(input("lado 3:")))
```

#### Exercício 3 

Você recebeu um dicionário com os nomes dos alunos e suas respectivas notas. Escreva uma função que calcula a média da turma e retorna uma lista com os nomes dos alunos que tiveram nota acima da média.

```python
# Solução
students = {
    "nomes": ['ana', 'joão', 'jéssica', 'luiz'],
    "notas": [10, 8, 3, 5],
    "media": 5
}

def main(students):
    avg = sum(students["notas"]) / len(students["notas"])
    print(f"Média da sala: {avg}")

    names = students["nomes"]
    grades = students["notas"]
    med_number = students["media"]

    names_med = []

    for i in range(0, len(names)):
        if grades[i] > med_number:
            names_med.append(names[i])

    print(f"Nomes c/ notas acima da média: {names_med}")


main(students)
```

#### Exercício 4

Dada uma lista de números, crie uma nova lista usando list comprehension que contenha o quadrado de cada número par da lista original.

```python
# Solução

lista = [1,2,3,4,5,6]

quadrados = list(map(lambda x: x**2, lista))

quadrados_pares = list(filter(lambda x: x%2==0, quadrados))


print(quadrados)
print(quadrados_pares)
```

#### Exercício 6

Você tem uma lista de dicionários, onde cada dicionário representa uma pessoa com nome e idade. Use a função sorted() com uma expressão lambda como chave (key) para ordenar a lista de pessoas da mais nova para a mais velha.

```python

pessoas = [
    {"nome": "Ana", "idade": 28},
    {"nome": "Bruno", "idade": 34},
    {"nome": "Carla", "idade": 22},
    {"nome": "Diego", "idade": 41},
    {"nome": "Fernanda", "idade": 30}
]



# Solução Propria

idades = []

for pessoa in pessoas:
    idade = pessoa["idade"]
    idades.append(idade)
    
idades_ordenadas = sorted(idades)
print(idades_ordenadas)

# Solução GPT

pessoas_ordenadas = sorted(pessoas, key=lambda pessoa: pessoa["idade"])

for pessoa in pessoas_ordenadas:
    print(f"Nome: {pessoa['nome']} - Idade: {pessoa['idade']}")
```

#### Exercício 8

Crie uma função que receba uma lista de strings (potenciais e-mails) e um parâmetro opcional dominio_desejado (com valor padrão "gmail.com"). A função deve retornar uma nova lista, usando list comprehension, contendo apenas os e-mails que terminam com o domínio desejado.

```python
# Solução

_emails = [
    "joao.silva@gmail.com",
    "maria.oliveira@hotmail.com",
    "carlos.santos@uol.com",
    "ana.pereira@gmail.com",
    "lucas.rocha@outlook.com"
]

def verificacao(emails, dominio_desejado = "gmail.com"):
    gmails = [e for e in emails if dominio_desejado in e]

    return gmails

verificacao(_emails)
```

#### Exercício 10

Crie um jogo onde o computador escolhe um número secreto entre 1 e 20. O jogador tem 5 tentativas para adivinhar. A cada tentativa, o programa informa se o palpite foi muito alto ou muito baixo. Se o jogador acertar, o loop deve ser interrompido com uma mensagem de vitória.

```python
# Solução

import random

numero_aleatorio = random.randint(1, 20) 
tentativas = 5
perimetro = 2
lista = []

lista.append(numero_aleatorio)

def intervalo(_n, _redondezas):
    for i in range(_redondezas):
        x = i+1
        y = _n - x
        lista.append(y)

    for i in range(_redondezas):
        x = i+1
        y = _n + x
        lista.append(y)
      
intervalo(numero_aleatorio, perimetro)

# print(f"numero_aleatorio: {numero_aleatorio}")  ###apagar depois
# print(f"lista: {lista}")

for i in range(tentativas):
    print(f"\ntentativa {i+1}")

    x = float(input("digite um número:"))
    if x == numero_aleatorio:
        print(f"Parabéns!!!\nVocê digitou: {x}\nVocê acertou o número aleatório ({numero_aleatorio})!!!")
        break

    elif x in lista:
        print(f"\nVocê digitou: {x}\nÓtimo palpite mas ainda não acertou...\nTente novamente!!!")

    elif x not in lista:
        if i == tentativas-1:
            print(f"\nVocê digitou: {x}\nInfelizmente você perdeu o jogo...o número era '{numero_aleatorio}'\nJogue novamente!!!")

        else:
            print(f"\nVocê digitou: {x}\nPassou longe colega...\nTente novamente!!!")
```
